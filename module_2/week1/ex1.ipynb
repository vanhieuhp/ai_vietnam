```python
from mimetypes import add_type

import matplotlib.pyplot as plt
import numpy as np
import matplotlib.image as mpimg
```

```python
img = mpimg.imread("module_2/tree.jpg")
plt.imshow(img)
```

```python
# max(rgb) + min(rgb) / 2
def color2grayscale(vector):
    return np.max(vector) * 0.5 + np.min(vector) * 0.5
```

```python
grayscale_01 = np.apply_along_axis(color2grayscale, 2, img)
print(grayscale_01.shape)
```

```python
plt.imshow(grayscale_01, cmap="gray")
```

```python
def color2grayscale(vector):
    return np.sum(vector)/3

grayscale_02 = np.apply_along_axis(color2grayscale, 2, img)
print(grayscale_02.shape)
```

```python
grayscale_03 = img.mean(axis=2)
print(grayscale_03.shape)
plt.imshow(grayscale_03, cmap="gray")
```
