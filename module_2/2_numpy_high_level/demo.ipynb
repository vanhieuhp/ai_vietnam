```python
import glob

# get paths
paths = glob.glob("samples/*.jpg")
paths = sorted(paths)

# resize
result = []
for path in paths:
    img = cv2.imread(path, 1)
    img = cv2.resize(img, (100, 100))
    result.append(img)

# convert list into array
result = np.array(result)
print(result.shape)
```
