```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('temperature-1d.csv').to_numpy()
temp = data[:,1]
```

```python
print(temp[:10])
```

```python
import matplotlib.pyplot as plt
```

```python
print(temp.shape)
```

```python
temp_day = temp.reshape(-1, 24)
print(temp_day.shape)
```

```python
mean_day = temp_day.mean(axis=1)
print(mean_day)
mean_day = mean_day.repeat(24)
print(mean_day.shape)
```

```python
plt.plot(temp, label='temp')
plt.plot(mean_day, label='mean_day')
plt.legend()
```
