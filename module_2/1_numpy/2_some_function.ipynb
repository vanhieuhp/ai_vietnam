```python
import numpy as np

shape = (4,5)
data = np.zeros(shape)
print(data)
```

```python
data = np.ones(shape)
print(data)
```

```python
print(data * 9)
```

```python
data = np.arange(0,5,0.5)
print(data)
```

```python
data = np.arange(10)
print(data.shape)

data = np.reshape(data, (2,5))
print(data)
```

```python
data = np.arange(10).reshape(2,5)
print(data.shape)

data = data.reshape(5, -1)
print(data)
```

```python
data = np.random.randint(1,5,(5,7))
print(data)
```

```python
data = np.arange(100).reshape(10,10)
print(data)
```

```python
data = np.array([1,4])
print(data)

data = np.repeat(data, 2)
print(data)
```

```python
data = np.arange(4).reshape(2,2)
print(data)

data = np.repeat(data, 2, axis=0)
data = np.repeat(data, 2, axis=1)
print(data)
```

```python
data = np.arange(9).reshape(3,3)
print(data)

sub_data = data
```

```python
print(data[1,:])
sub_data = data[1,:]
sub_data[2] = 20
print(sub_data)
```

```python
data = np.arange(4).reshape(2,2)
sub_data = data[:,:]
sub_data = 20
print(data)
```

```python
data = np.arange(9).reshape(3,3)
print(data)
```

```python
print(data[0,0])
print(data[1,2])

print(data[[0,1], [0,2]])
print(data[[1,2], [0,2]])
```

```python
data = np.arange(9).reshape(3,3)
print(data)
print(np.sum(data))
```

```python
data = np.arange(4).reshape(2,2)
print(data)
print(np.sum(data, axis=0))
```

```python
data = np.array([1,2,3])
print(data)

max_value = np.max(data)
print(max_value)
```

```python
data_exp = np.exp(data)
print(data_exp)
sum_data_exp = np.sum(data_exp)
print(sum_data_exp)
```

```python
result = data_exp / sum_data_exp
print(result)
```
