```python
import cv2
import numpy as np

# 1. Đọc 3 ảnh và resize về cùng kích thước
size = (640, 360)
B = cv2.resize(cv2.imread("module_2/background/B.jpg"), size)
I = cv2.resize(cv2.imread("module_2/background/I.jpg"), size)
F = cv2.resize(cv2.imread("module_2/background/F.jpg"), size)

# 2. Tính độ lệch tuyệt đối D = |I - B|
# Cách an toàn: dùng cv2.absdiff (hoặc np.abs(I.astype(np.int16) - B.astype(np.int16)))
D = cv2.absdiff(I, B)

# Lấy trung bình độ lệch trên 3 kênh màu (B, G, R) -> ảnh 2D (H, W)
D_mean = np.mean(D, axis=2)

# 3. Tạo mask theo ngưỡng (ảnh AI tạo có thể lệch sáng nhẹ, hãy thử chỉnh 15 -> 40)
threshold = 25
mask = (D_mean > threshold)[..., None]  # mở rộng thành (H, W, 1) để broadcast 3 kênh

# 4. Ghép ảnh O: chỗ nào là người lấy I, chỗ nào là nền lấy F
O = np.where(mask, I, F).astype(np.uint8)

cv2.imwrite("module_2/background/O_output.png", O)
```
